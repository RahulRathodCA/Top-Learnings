```python
import cv2, numpy as np, matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()
img = cv2.imread(list(uploaded.keys())[0])
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)


H, S, V = hsv[img.shape[0]//2, img.shape[1]//2]
lower = np.array([max(H-10,0), max(S-50,0), max(V-50,0)]) #hue(0-179),#sat(0-255),#val()
upper = np.array([min(H+10,179), min(S+50,255), min(V+50,255)])


mask = cv2.inRange(hsv, lower, upper)
result = cv2.bitwise_and(img, img, mask=mask)

# Show
plt.subplot(1,2,1); plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); plt.axis('off')
plt.subplot(1,2,2); plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB)); plt.axis('off')
plt.show()

```

```python
from google.colab.patches import cv2_imshow
import cv2
import numpy as np #numerical python
from google.colab import files

uploaded = files.upload()
can = list(uploaded.keys())[0]
img = cv2.imread(can)

def edge_detect(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 500, 50) #canny edge detection #density
    return cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)

def blur_image(image):
    return cv2.GaussianBlur(image, (90,40), 0)

def cartoonize(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    gray = cv2.medianBlur(gray, 5)
    edges = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C,
                                  cv2.THRESH_BINARY, 9, 9)
    color = cv2.bilateralFilter(image, 9, 300, 300)
    cartoon = cv2.bitwise_and(color, color, mask=edges)
    return cartoon

def invert_colors(image):
    return 255 - image

while True:
    print("\nChoose an option:")
    print("1 - Show Normal Image")
    print("2 - Edge Detection")
    print("3 - Blur Image")
    print("4 - Cartoonize")
    print("5 - Invert Colors")


    choice = input("Entered choice is : ")

    if choice == "1":
        cv2_imshow(img)
        break
    elif choice == "2":
        cv2_imshow(edge_detect(img))
        break
    elif choice == "3":
        cv2_imshow(blur_image(img))
        break
    elif choice == "4":
        cv2_imshow(cartoonize(img))
        break
    elif choice == "5":
        cv2_imshow(invert_colors(img))
        break
    else:
        print("Invalid choice!")
        break
 #note: by using thonny or else any other interface tools we can easily create the user interface
```
